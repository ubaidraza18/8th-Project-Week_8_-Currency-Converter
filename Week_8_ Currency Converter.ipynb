```python
import requests

print("\n============== Currency Converter ==============")

try:
    amount = float(input("Enter Amount: "))
    from_currency = input("Enter From Currency (eg. USD, EUR, GBP): ").strip().upper()
    to_currency = input("Enter To Currency (eg. USD, EUR, GBP): ").strip().upper()

    if from_currency == "" or to_currency == "":
        print("From Currency and To Currency cannot be empty.")

    elif amount < 0:
        print("Please Enter Amount Greater Than or Equal to 0.")

    elif from_currency == to_currency:
        print("From Currency and To Currency cannot be the same.")

    else:
        # API Request to Get Exchange Rate
        url = f"https://api.frankfurter.dev/v2/rate/{from_currency}/{to_currency}"
        response = requests.get(url , timeout=10)

        if response.status_code == 200:
            data = response.json()
            exchange_rate = data["rate"]
            converted_amount = amount * exchange_rate
            print("\n==================== Conversion Result ====================")
            print(f"Exchange Rate: 1 {from_currency} = {exchange_rate:.4f} {to_currency}")
            print(f"{amount} {from_currency} = {converted_amount:.2f} {to_currency}")
        else:
            print("Error occurred while fetching exchange rate.")

except ValueError:
    print("Please Enter Amount As A Number.")

except requests.exceptions.RequestException:
    print("An error occurred while making the request. Please check your internet connection and try again.")
```
